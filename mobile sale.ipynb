```python
if df['customer_ratings'].isnull().sum() > 0:
    df['customer_ratings'] = (
        df.groupby('brand')['customer_ratings']
          .transform(lambda x: x.fillna(x.median()))
    )
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python
df.columns
```

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group_check'] = pd.qcut(df['customer_age'], q=4, labels=labels)
df[['customer_age', 'age_group', 'age_group_check']].head(10)
```

```python
same_date = (
    pd.to_datetime(dict(year=df.year, month=df.month, day=df.day)) == df['purchase_date']
).all()
print('day/month/year fully redundant with purchase_date:', same_date)
 
if same_date:
    df = df.drop(columns=['day', 'month', 'year'])
```

```python
df.columns
```

```python
calc_total = (df['units_sold'] * df['price_per_unit_usd']).round(2)
match_rate = (calc_total == df['total_sale_amount']).mean()
print(f'total_sale_amount matches units_sold * price_per_unit_usd: {match_rate:.1%} of rows')
```
